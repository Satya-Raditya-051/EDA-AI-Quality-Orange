# EDA

**Exploratory Data Analysis** untuk data tabular adalah proses menganalisis dan memahami struktur, pola, dan karakteristik data dalam bentuk tabel (seperti CSV, Excel, atau database). EDA sangat penting sebelum melakukan pemodelan atau analisis lebih lanjut karena membantu mengidentifikasi masalah data, memahami distribusi, dan menemukan hubungan antar variabel.

```python
# checking duplicate
data_nd = data.drop_duplicates(subset = "Variety")
data_nd
```

```python
# checking balance label
data.value_counts("Variety")
```

```python
# visualisi data dari tiap label
plt.figure(figsize=(12, 6))  # Perbesar ukuran figure
sns.countplot(x='Variety', data=data)
plt.xticks(rotation=45, ha="right")  # Rotasi label agar lebih jelas
plt.show()
```

```python
# Relation between variables
sns.scatterplot(x='Size (cm)', y='Weight (g)',
                hue='Variety', data=data, )
 
# Placing Legend outside the Figure
plt.legend(bbox_to_anchor=(1, 1), loc=2)
 
plt.show()
```

```python
# plot all the columns' relationship
sns.pairplot(data.drop(['Size (cm)'], axis=1, errors='ignore'), 
             hue='Variety', height=2)
plt.show()
```

```python

# Plot histograms for all numerical columns
data.hist(figsize=(10, 20), bins=20, edgecolor='black')

plt.suptitle("Histograms of All Numerical Columns", fontsize=14)
plt.show()
```

```python
# Histogram with Displot
plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3) 
plot.map(sns.distplot, "Size (cm)").add_legend()
 
plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3)
plot.map(sns.distplot, "Weight (g)").add_legend()
 
plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3)
plot.map(sns.distplot, "Brix (Sweetness)").add_legend()
 
plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3)
plot.map(sns.distplot, "pH (Acidity)").add_legend()

plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3)
plot.map(sns.distplot, "Softness (1-5)").add_legend()

plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3)
plot.map(sns.distplot, "HarvestTime (days)").add_legend()

plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3)
plot.map(sns.distplot, "Ripeness (1-5)").add_legend()

plot = sns.FacetGrid(data, hue="Variety",height=5, aspect=3)
plot.map(sns.distplot, "Quality (1-5)").add_legend()
 
plt.show()
```

```python
numeric_data = data.select_dtypes(include=[np.number])
correlation_matrix = numeric_data.corr(method='pearson')
print(correlation_matrix)
```
